## Phase 1 — Python Basics

1. VARIABLES

Variable ek naam hai jiske through hum kisi value/object ko refer karte hain.

Example:

```python
name = "Python"
age = 20
price = 99.5

print(name)
print(age)
print(price)
```

2. BASIC DATA TYPES

Python mein alag-alag tarah ka data store karne ke liye built-in data types hote hain:

Integers (int): Bina decimal waale numbers (e.g., 10, -5).
Floats (float): Decimal waale numbers (e.g., 3.14, 10.5).
Strings (str): Text ya words jinhe quotes (" " ya '') mein likha jata hai.
Booleans (bool): Sirf do values - True ya False.
Complex Numbers: Python mein format a + bj hota hai.

```python
x = 10
y = 2.5
text = "Python"
is_active = True
z = 2 + 3j

print(type(x))
print(type(y))
print(type(text))
print(type(is_active))
print(type(z))
```

3. INPUT AND OUTPUT

input() user se value lene ke liye use hota hai.
print() output screen par dikhane ke liye use hota hai.

```python
name = input("Enter your name: ")
print("Hello", name)
```

## Phase 2 — Type Conversion / Type Casting

4. TYPE CONVERSION VS TYPE CASTING

Type Conversion (Implicit):
Jab Python khud hi ek data type ko doosre mein badal deta hai
taaki data loss na ho.

```python
x = 10
y = 2.5
result = x + y
print(result)
```

Type Casting (Explicit):
Jab hum programmer manually kisi data type ko forcefully change karte hain
(e.g., int(), str(), float()).

```python
num_str = "100"
num_int = int(num_str)
print(num_int + 50)

print(str(100))
print(float(10))
print(int(2.9))
```

5. ord() FUNCTION

ord() function kisi bhi single character ka ASCII value
(ya Unicode code point) return karta hai.

```python
print(ord("A"))  # Output: 65
print(ord("a"))  # Output: 97
```

## Phase 3 — Operators

6. ARITHMETIC OPERATORS

+, -, *, /, //, %, **

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

7. COMPARISON OPERATORS

>, <, >=, <=, ==, !=

Comparison ka result True ya False hota hai.

```python
print(10 > 5)
print(10 == 10)
print(10 != 5)
```

8. ASSIGNMENT OPERATORS

=, +=, -=, *=, /=, //=, %=, **=

```python
number = 10
number += 5
print(number)
```

9. LOGICAL OPERATORS

and, or, not

```python
print(10 > 5 and 20 > 10)
print(10 > 5 or 5 > 20)
print(not True)
```

10. OPERATOR PRECEDENCE

High-priority waala operator pehle evaluate hota hai.

```python
result = 10 + 5 * 2
print(result)
```

Yahan * (multiplication) ki priority + (addition) se high hai.

Pehle 5 * 2 = 10 hoga, phir 10 + 10 = 20 hoga.

Agar aap + pehle karwana chahte ho, toh Parentheses () use karo,
kyunki unki priority sabse high hoti hai.

```python
result = (10 + 5) * 2
print(result)
```

11. ASSOCIATIVITY

Jab do operators ki precedence bilkul same hoti hai,
tab direction dekhi jaati hai.

A. Left-to-Right Associativity
Normal operators jaise +, -, *, / left se right taraf calculate hote hain.

```python
result = 100 / 10 / 2
print(result)
```

Pehle left waala solve hoga: 100 / 10 = 10.0
Phir right waala: 10.0 / 2 = 5.0

B. Right-to-Left Associativity
Exponentiation (**) yani power operator Right-to-Left chalta hai.

```python
result = 2 ** 3 ** 2
print(result)
```

## Phase 4 — Strings

12. STRINGS AND QUOTES

Python mein teen tarah ke quotes se string banayi ja sakti hai,
koi alag character data type nahi hota (sab str hain).

Single ('a') & Double ("a"):
Single-line strings ke liye use hote hain. Standard text ke liye dono same hain.

Triple quotes:
Multi-line strings aur docstrings ke liye use hote hain.

```python
s1 = "a"
s2 = 'a'
s3 = """Yeh multi-line
string hai."""

print(s1)
print(s2)
print(s3)
```

13. STRING INDEXING

String ka har character index par hota hai.
Indexing zero-based hoti hai.

```python
name = "Python"

print(name[0])
print(name[2])
print(name[-1])
```

14. STRING SLICING

Slicing se string ka ek portion nikala ja sakta hai.

```python
print(name[0:3])
print(name[2:])
print(name[:4])
```

15. STRING IMMUTABILITY

Python mein String immutable hoti hai, yani ek baar banne ke baad
uske andar ke kisi character ko direct change nahi kiya ja sakta.

```python
name = "Python"

# name[0] = 'J'
# TypeError aayega!

# Agar badalna hai, toh nayi string banani padti hai:
name = "J" + name[1:]
print(name)
```

16. STRING METHODS & OPERATIONS

Python mein String class (str) ke paas bohot se useful built-in methods hote hain.

```python
text = "  python programming  "

print(text.upper())
print(text.lower())
print(text.strip())
print(text.replace("python", "Java"))
print(text.find("pro"))
```

## Phase 5 — Decision Making

17. BOOLEAN CONDITIONS

Conditions ka result True ya False hota hai.

```python
age = 20
print(age >= 18)
```

18. IF / ELSE / ELIF

Decision constructs (if-elif-else) code ke execution flow ko
condition ke hisab se control karte hain.

```python
age = 18

if age < 18:
    print("Minor")
elif age == 18:
    print("Just became adult")
else:
    print("Adult")
```

19. NESTED IF

Ek if ke andar doosra if bhi use kar sakte hain.

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Allowed")
```

20. MATCH-CASE

Python mein traditional switch-case nahi hota, par Python 3.10+
mein match-case ka feature diya gaya hai jo switch-case jaisa
hi kaam karta hai.

```python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
```

## Phase 6 — Loops

21. WHILE LOOP

Jab tak condition True hai, loop chalta rahega.

```python
i = 0

while i < 5:
    print(i)
    i += 1
```

22. FOR LOOP

Sequence ke elements par iterate karne ke liye for loop use hota hai.

```python
for i in range(5):
    print(i)
```

23. range() FUNCTION

range() sequence of numbers generate karta hai.
Syntax: range(start, stop, step)

```python
for i in range(5):
    print(i)

print(list(range(1, 10, 2)))
```

24. BREAK

Yeh loop ko wahi rok kar turant bahar nikal deta hai.

```python
for i in range(5):
    if i == 3:
        break
    print(i)
```

25. CONTINUE

Current iteration ko skip karke next iteration par chala jata hai.

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

26. PASS

Yeh ek empty placeholder (null statement) hai.
Jab aapko future code ke liye koi block khali chhodna ho
taaki syntax error na aaye, tab pass use karte hain.

```python
if True:
    pass
```

27. NESTED LOOPS

Ek loop ke andar doosra loop.

Pattern questions ke liye nested loops bahut important hain.

```python
for i in range(3):
    for j in range(3):
        print("*", end="")
    print()
```

## Phase 7 — Lists

28. LIST INTRODUCTION

Python mein native arrays ki jagah aamtaar par Lists ka use hota hai.

```python
my_list = ["Python", "Java", "C++", "JavaScript"]
print(my_list)
```

29. LIST INDEXING

Indexing zero-based hoti hai.
Pehla element 0 position par hota hai.

```python
print(my_list[0])
print(my_list[2])
print(my_list[-1])
```

30. LIST SLICING

```python
print(my_list[0:2])
print(my_list[1:])
```

31. UPDATE LIST ELEMENT

```python
my_list[1] = "SQL"
print(my_list)
```

32. ADD ELEMENTS

```python
my_list.append("Azure")
my_list.insert(1, "Python")
print(my_list)
```

33. REMOVE ELEMENTS

```python
my_list.remove("SQL")
removed = my_list.pop()
print(my_list)
print(removed)
```

34. LOOP THROUGH LIST

```python
fruits = ["Apple", "Banana", "Cherry"]

for fruit in fruits:
    print(fruit)
```

35. ENUMERATE()

Agar aapko loop mein element aur uska index dono ek saath chahiye,
toh enumerate() function sabse best tarika hai.

```python
for index, fruit in enumerate(fruits):
    print(f"Index {index} par {fruit} hai")
```

## Phase 8 — Other Python Collections

36. TUPLE

Tuple ordered collection hai aur generally immutable hota hai.

```python
numbers = (10, 20, 30)
print(numbers[0])
```

37. SET

Set unique values store karta hai.
Duplicate values automatically remove ho jaati hain.

```python
values = {1, 2, 2, 3, 4}
print(values)
```

38. DICTIONARY

Dictionary data ko key-value pair mein store karta hai.

```python
student = {
    "name": "Yogesh",
    "age": 20
}

print(student["name"])
print(student["age"])
```

39. NESTED COLLECTIONS

Collections ko doosri collections ke andar bhi store kar sakte hain.

```python
students = [
    {"name": "A", "age": 20},
    {"name": "B", "age": 21}
]

print(students[0]["name"])
```

## Phase 9 — Membership & Useful Built-In Functions

40. MEMBERSHIP OPERATORS

Yeh check karne ke liye use hote hain ki koi value kisi sequence
(list, string, tuple) mein exist karti hai ya nahi.

in: True deta hai agar value milti hai.
not in: True deta hai agar value nahi milti.

```python
fruits = ["apple", "banana", "mango"]

print("banana" in fruits)
print("grape" not in fruits)

print("Py" in "Python")
```

41. USEFUL BUILT-IN FUNCTIONS

```python
numbers = [10, 5, 20, 15]

print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
print(sorted(numbers))
```

42. ZIP()

Multiple sequences ko pair/group karne ke liye use hota hai.

```python
names = ["A", "B", "C"]
marks = [80, 90, 85]

for name, mark in zip(names, marks):
    print(name, mark)
```

## Phase 10 — Bitwise Operators

43. BINARY BASICS

Bitwise operators numbers ke binary bits (0s aur 1s) par kaam karte hain.

Important operators:
&  AND
|  OR
^  XOR
~  NOT
<< Left Shift
>> Right Shift

```python
a = 5
b = 3

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
```

44. LEFT SHIFT (<<)

Bits ko left ki taraf shift karta hai aur right mein zeros add karta hai.
Formula: x * 2^n

```python
a = 5
result = a << 2
print(result)
```

45. RIGHT SHIFT (>>)

Bits ko right ki taraf shift karta hai aur right side ke bits discard
kar deta hai.
Formula: x // 2^n

```python
a = 20
result = a >> 2
print(result)
```

## Phase 11 — Functions

46. FUNCTION

Function reusable block of code hota hai.

```python
def greet():
    print("Hello Python")

greet()
```

47. PARAMETERS AND ARGUMENTS

```python
def greet_user(name):
    print("Hello", name)

greet_user("Yogesh")
```

48. RETURN

return function se value wapas bhejne ke liye use hota hai.

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

49. DEFAULT ARGUMENT

```python
def greet(name="User"):
    print("Hello", name)

greet()
greet("Yogesh")
```

50. KEYWORD ARGUMENT

```python
def student_info(name, age):
    print(name, age)

student_info(age=20, name="Yogesh")
```

51. *args

Multiple positional arguments ko receive karne ke liye use hota hai.

```python
def total(*numbers):
    return sum(numbers)

print(total(10, 20, 30))
```

52. **kwargs

Multiple keyword arguments ko receive karne ke liye use hota hai.

```python
def show_info(**info):
    print(info)

show_info(name="Yogesh", age=20)
```

53. SCOPE

Local variable function ke andar available hota hai.
Global variable broader scope mein available hota hai.

```python
x = 100

def demo():
    y = 20
    print(x)
    print(y)

demo()
```

54. MUTABLE VS IMMUTABLE / OBJECT-REFERENCE

Python mein Object-Reference pass hota hai.
Function ke andar modification is baat par depend karti hai
ki object mutable hai ya immutable.

Strings immutable hoti hain, jabki lists mutable hoti hain.

```python
def change_list(items):
    items.append(100)

numbers = [1, 2, 3]
change_list(numbers)

print(numbers)


def change_string(value):
    value = value + " World"

text = "Hello"
change_string(text)

print(text)


# QUICK REVISION ROADMAP
```
