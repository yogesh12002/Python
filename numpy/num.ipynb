```python
import numpy as np
from pathlib import Path

# Use the notebook's current working directory, and fall back to the known folder
csv_path = Path.cwd() / "NumPy_Retail_Sales_Practice_Dataset.csv"
if not csv_path.exists():
    csv_path = Path(r"C:\Users\ygupt\Desktop\python\numpy") / "NumPy_Retail_Sales_Practice_Dataset.csv"

if not csv_path.exists():
    candidates = [
        csv_path,
        Path.cwd() / "NumPy_Retail_Sales_Practice_Dataset.csv",
        Path.cwd().parent / "NumPy_Retail_Sales_Practice_Dataset.csv",
        Path(r"C:\Users\ygupt\Desktop\python\numpy") / "NumPy_Retail_Sales_Practice_Dataset.csv",
    ]
    csv_path = next((p for p in candidates if p.exists()), csv_path)

    if not csv_path.exists():
        filename = "NumPy_Retail_Sales_Practice_Dataset.csv"
        search_roots = [
            Path.cwd(),
            Path.cwd().parent,
            Path(r"C:\Users\ygupt\Desktop\python"),
        ]

        matches = []
        for root in search_roots:
            if root.exists():
            matches.extend(
                p for p in root.rglob("*")
                if p.is_file() and p.name.lower() == filename.lower()
            )

        if matches:
            csv_path = matches[0]
        else:
            raise FileNotFoundError(
            f"Could not find {filename}. "
            "Place the CSV in the notebook folder or update its filename/path."
            )

data = np.genfromtxt(
    str(csv_path),
    delimiter=',',
    names=True,
    dtype=None,
    encoding='utf-8',
    autostrip=True
)

# Access columns directly by header name:
print(data['Order_ID'])
print(data['Net_Amount'])
```
