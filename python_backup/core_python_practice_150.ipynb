- Use the topic(s) listed above rather than hard-coding the final answer.

```python
# Write your solution here.
# Add your own test cases below.
def bill(qty, unit_price, discount_pct=0, tax_pct=0):
    subtotal = qty * unit_price
    discounted = subtotal * (1 - discount_pct / 100)
    tax = discounted * tax_pct / 100
    return discounted + tax

print(bill(3, 500, 10, 18))

```
