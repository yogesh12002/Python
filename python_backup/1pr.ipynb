```python
name = input("Please enter your name: ")
age = int(input("Please enter your age: "))
city = input("Please enter your city: ")
address = input("Please enter your address: ")
print(f"Hello, {name}! You are {age} years old and live in {city}. Your address is {address}.")
```

```python
marks = int(input("Enter marks: "))
print(f"Marks: {marks}")
if marks>=90:
  print("Grade A")
elif marks>=80:
  print("Grade B")
elif marks>=70:
  print("Grade C")
else :
  print("Fail")
```
