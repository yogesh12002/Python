## Practice 4/5

**Problem:** Build a simple Caesar-shift encoder for alphabetic characters using ord() and chr().

**Acceptance criteria:**
- Use the topic(s) listed above rather than hard-coding the final answer.
- Test at least 2 normal cases and 1 edge case.
- Keep the solution readable and split repeated logic into functions when appropriate.


```python
# Solution
def caesar(text, shift):
    out = []
    for ch in text:
        if "A" <= ch <= "Z":
            out.append(chr((ord(ch)-ord("A")+shift)%26 + ord("A")))
        elif "a" <= ch <= "z":
            out.append(chr((ord(ch)-ord("a")+shift)%26 + ord("a")))
        else:
            out.append(ch)
    return "".join(out)

print(caesar("Hello, World!", 3))
```

## Practice 5/5

**Problem:** Build a simple feature-toggle system where individual features are represented by bits.

**Acceptance criteria:**
- Use the topic(s) listed above rather than hard-coding the final answer.
- Test at least 2 normal cases and 1 edge case.
- Keep the solution readable and split repeated logic into functions when appropriate.


```python
# Solution
FEATURES = {"dark_mode": 1, "notifications": 2, "analytics": 4, "beta": 8}
flags = FEATURES["dark_mode"] | FEATURES["analytics"]

def enabled(name):
    return bool(flags & FEATURES[name])

def set_feature(name, on=True):
    global flags
    if on:
        flags |= FEATURES[name]
    else:
        flags &= ~FEATURES[name]

print({name: enabled(name) for name in FEATURES})
set_feature("beta")
set_feature("analytics", False)
print({name: enabled(name) for name in FEATURES})
```

## Practice 3/5

**Problem:** Build a menu-driven program using match-case for CRUD-style choices.

**Acceptance criteria:**
- Use the topic(s) listed above rather than hard-coding the final answer.
- Test at least 2 normal cases and 1 edge case.
- Keep the solution readable and split repeated logic into functions when appropriate.


```python
# Solution
records = {}
def crud(choice, key=None, value=None):
    match choice:
        case "create":
            records[key] = value
        case "read":
            return records.get(key)
        case "update":
            if key in records: records[key] = value
        case "delete":
            records.pop(key, None)
        case _:
            return "Invalid choice"

crud("create", "P100", {"name":"Mouse"})
print(crud("read", "P100"))
crud("update", "P100", {"name":"Gaming Mouse"})
print(crud("read", "P100"))
crud("delete", "P100")
print(records)
```

## Practice 5/5

**Problem:** Build a retry decorator that reruns a function a limited number of times when it raises an exception.

**Acceptance criteria:**
- Use the topic(s) listed above rather than hard-coding the final answer.
- Test at least 2 normal cases and 1 edge case.
- Keep the solution readable and split repeated logic into functions when appropriate.


```python
# Solution
from functools import wraps
def retry(times=3):
    def deco(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, times+1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"attempt {attempt} failed:", e)
                    if attempt == times: raise
        return wrapper
    return deco

counter = {"n":0}
@retry(3)
def flaky():
    counter["n"] += 1
    if counter["n"] < 3: raise RuntimeError("temporary failure")
    return "success"
print(flaky())
```

```python
# Solution
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner, self.balance = owner, balance
    def deposit(self, amount):
        if amount <= 0: raise ValueError("amount must be positive")
        self.balance += amount
    def withdraw(self, amount):
        if amount <= 0 or amount > self.balance: raise ValueError("invalid withdrawal")
        self.balance -= amount
    def __repr__(self):
        return f"BankAccount({self.owner!r}, balance={self.balance:.2f})"

a = BankAccount("Yogesh", 1000)
a.deposit(500); a.withdraw(200)
print(a)
```

**Problem:** Build an inventory manager using OOP, dictionaries, file persistence, validation, and custom exceptions.

**Acceptance criteria:**
- Use the topic(s) listed above rather than hard-coding the final answer.
- Test at least 2 normal cases and 1 edge case.
- Keep the solution readable and split repeated logic into functions when appropriate.


```python
# Solution
import json
class OutOfStockError(Exception): pass
class Inventory:
    def __init__(self, data=None): self.data = data or {}
    def sell(self, pid, qty):
        if pid not in self.data: raise KeyError(pid)
        if qty > self.data[pid]: raise OutOfStockError(pid)
        self.data[pid] -= qty
    def save(self, path):
        with open(path,"w",encoding="utf-8") as f: json.dump(self.data,f)

inv = Inventory({"P100":10,"P200":5})
inv.sell("P100",3)
inv.save("inventory.json")
print(inv.data)
```
