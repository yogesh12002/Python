**Dictionary Basics & Nesting**

A dictionary is an unordered, mutable collection of key-value pairs written with curly braces {}.

Dictionary Keys vs Values: Keys must be immutable (hashable) types like strings, numbers, or tuples. Lists cannot be used as dictionary keys because they are mutable.

Nesting: A dictionary value can be another dictionary.

````python
# Nested Dictionary
student = {
    "name": "Rahul",
    "marks": {"math": 85, "science": 90}
}

# Accessing nested value
print(student["marks"]["math"])  # Output: 85
````

**Converting to Dictionary (dict() & Key Immutability)**

You can convert a collection of key-value pairs (like a list/tuple of 2-item tuples) into a dictionary using dict().

````python
# Conversion using tuples (valid because inner elements are key-value pairs)
pair_tuples = (("a", 1), ("b", 2))
d = dict(pair_tuples)  # Output: {'a': 1, 'b': 2}

# Why List CANNOT be a KEY:
# d = {[1, 2]: "value"}  --> TypeError: unhashable type: 'list'
````

**Adding/Updating Values (Setting Values)**

````python
person = {"name": "Alice"}
person["age"] = 25  # Adds a new key-value pair
person["name"] = "Bob"  # Updates existing key
print(person)  # Output: {'name': 'Bob', 'age': 25}
````

**Iterating through a Dictionary (Loops & .items())**

````python
data = {"a": 10, "b": 20, "c": 30}

# Print keys only
for key in data.keys():
    print(key)

# Print values only
for val in data.values():
    print(val)

# Print key-value pairs together using .items()
for key, val in data.items():
    print(f"{key}: {val}")
````

**Safe Access with .get()**

Using dict[key] raises a KeyError if the key doesn't exist, whereas .get(key, default) returns None (or a custom default message).

````python
user = {"name": "Amit"}
print(user.get("age"))  # Output: None (no error)
print(user.get("age", 18))  # Output: 18 (default value used)
````

**Deleting Key-Value Pairs**

del dict[key]: Removes key-value pair or deletes the entire dictionary variable.

pop(key): Removes the key and returns its value.

popitem(): Removes and returns the last inserted key-value pair as a tuple (key, value).

clear(): Removes all items, leaving an empty dictionary {}.

````python
info = {"a": 1, "b": 2, "c": 3, "d": 4}

del info["a"]  # Deletes 'a'
val = info.pop("b")  # Removes 'b' and stores 2 in val
last = info.popitem()  # Removes ('d', 4)

print(info)  # Output: {'c': 3}

info.clear()  # Empties the dictionary
print(info)  # Output: {}
````

**Special Methods (fromkeys & zip)**

.fromkeys(seq, value): Creates a dictionary with keys from a sequence and sets all values to a default.

zip(keys, values): Pairs elements from two lists/tuples together to form a dictionary.

````python
# fromkeys
keys = ["name", "age", "city"]
new_dict = dict.fromkeys(keys, "Unknown")
print(new_dict)  # Output: {'name': 'Unknown', 'age': 'Unknown', 'city': 'Unknown'}

# zip
k = ["x", "y", "z"]
v = [1, 2, 3]
zipped_dict = dict(zip(k, v))
print(zipped_dict)  # Output: {'x': 1, 'y': 2, 'z': 3}
````

**Dunder Methods & Specialization Methods**

Dunder (Double UNDERscore) or magic methods allow custom classes to emulate built-in behaviors (like operators or iteration):

````python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):  # Called by print()
        return f"'{self.title}' with {self.pages} pages"

    def __len__(self):  # Called by len()
        return self.pages

b = Book("Python 101", 300)
print(b)        # Output: 'Python 101' with 300 pages
print(len(b))   # Output: 300
````

**Anonymous Functions (lambda), map, filter, and reduce**

Why use lambda? It creates small, single-expression inline functions without formal def declarations. You can call them directly or via variable references.

map(): Applies a function to all elements in an iterable.

filter(): Filters elements based on a boolean function.

reduce(): Cumulatively reduces an iterable to a single value (requires functools).

````python
from functools import reduce

nums = [1, 2, 3, 4, 5]

# map with lambda: Square each number
squared = list(map(lambda x: x**2, nums))  # [1, 4, 9, 16, 25]

# filter with lambda: Keep even numbers
evens = list(filter(lambda x: x % 2 == 0, nums))  # [2, 4]

# reduce with lambda: Sum all elements
total = reduce(lambda x, y: x + y, nums)  # 15
````

**Difference Between Expression and Method**

Expression: A piece of code that evaluates to a single value (e.g., x + 5, len("hi"), lambda x: x*2).

Method: A function that belongs to an object or class, invoked using dot notation (e.g., "hello".upper(), my_list.append(1)).

**Iterators, Generators, and Decorators**

Iterator: An object with state that yields items sequentially using __iter__() and __next__().

Generator: A function that yields a sequence of values lazily using yield instead of storing them all in memory.

Decorator: A wrapper function that modifies or extends the behavior of another function without changing its source code.

**iter() and next():** Convert an iterable into an iterator and fetch items sequentially.

````python
lst = [10, 20]
it = iter(lst)
print(next(it))  # Output: 10
print(next(it))  # Output: 20
````

````python
# 1. Generator Example
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

gen = count_up_to(3)
print(next(gen))  # Output: 1
print(next(gen))  # Output: 2

# 2. Decorator Example
def my_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper

@my_decorator
def say_hi():
    print("Hi!")

say_hi()
````

**Generators & yield:** Functions that yield values one at a time, preserving execution state between calls.

````python
def count_up(n):
    for i in range(1, n + 1):
        yield i
gen = count_up(3)
print(next(gen))  # Output: 1
````

**Variable Scope & Namespace**

keywords: Reserved words in Python (e.g., def, yield, import, nonlocal).

locals(): Returns a dictionary representing the current local namespace/variables.

globals(): Returns a dictionary representing the current global namespace/variables.

nonlocal: Declares that a variable inside a nested function belongs to the outer enclosing function's scope.

````python
def outer():
    x = "initial"
    def inner():
        nonlocal x
        x = "modified"
    inner()
    return x
print(outer())  # Output: modified
````

**Nested Functions & Closures**

Nested Functions (Method inside Method): Defining a function inside another function to encapsulate behavior.

Closure Properties: A inner function retains access to variables from its enclosing lexical scope even after the outer function has finished executing.

````python
def multiplier(factor):
    def multiply(number):
        return number * factor  # Remembers 'factor'
    return multiply

double = multiplier(2)
print(double(5))  # Output: 10
````
