# 1. Python Basics & Object References

**Focus:** Variables and object references, mutable vs immutable behavior, type() and id(), basic built-in types, function argument behavior

### Key ideas
- A variable is a **name bound to an object**.
- `type(x)` tells you the object's type.
- `id(x)` identifies the object during its lifetime.
- Mutable objects can be changed in place; immutable objects cannot.
- Python passes object references to functions (often described as *pass-by-object-reference*).

### Syntax
```python
name = value
type(name)
id(name)
isinstance(name, SomeType)
```

### Mutable vs immutable
```python
items = [10, 20]
items.append(30)       # same list object is changed

x = 10
x = x + 1              # a new integer object is bound to x
```

Common mutable types: `list`, `dict`, `set`.
Common immutable types: `int`, `float`, `bool`, `str`, `tuple`.

### Function argument behavior
```python
def update(data, number):
    data.append("new")   # mutates caller's list
    number += 1          # rebinds local name

values = []
n = 5
update(values, n)
```

**How to solve the exercises:** track the object and the variable separately. Ask: “Did I mutate the object, or did I reassign the variable?”

# 24. Closures, nonlocal & Function Factories

**Focus:** nested functions, closure state, nonlocal, function factories, encapsulation

### Closure
A closure is an inner function that remembers values from its enclosing scope.

```python
def make_counter():
    count = 0

    def counter():
        nonlocal count
        count += 1
        return count

    return counter

c = make_counter()
print(c())
print(c())
```

### Function factory
```python
def make_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply
```

**How to solve:** identify the state that must remain alive after the outer function returns. Store it in the enclosing scope and use `nonlocal` when you need to modify it.


# 25. Decorators

**Focus:** basic decorators, decorators with args, *args/**kwargs wrappers, functools.wraps, decorator chaining

### Basic decorator
```python
from functools import wraps

def log_call(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Calling:", func.__name__)
        result = func(*args, **kwargs)
        print("Done")
        return result
    return wrapper
```

Use:
```python
@log_call
def add(a, b):
    return a + b
```

### Decorator with configuration
```python
def repeat(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            result = None
            for _ in range(times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator
```

**How to solve:** think of a decorator as `new_function = decorator(old_function)`. Preserve arguments with `*args, **kwargs` and preserve metadata with `@wraps`.

# 31. Dunder & Special Methods

**Focus:** __str__, __repr__, __len__, __eq__, __add__, __iter__

Dunder methods define how objects work with Python's built-in operations.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return self.name

    def __repr__(self):
        return f"Product({self.name!r}, {self.price!r})"

    def __len__(self):
        return len(self.name)

    def __eq__(self, other):
        return isinstance(other, Product) and self.name == other.name

    def __add__(self, other):
        return self.price + other.price
```

### Iteration
```python
def __iter__(self):
    return iter(self.items)
```

**How to solve:** map the operation to the special method: `str(obj)` → `__str__`, `len(obj)` → `__len__`, `a == b` → `__eq__`, `a + b` → `__add__`.

### Question 4/5

**Problem:** Build a simple Caesar-shift encoder for alphabetic characters using ord() and chr().

**Write the complete Python solution in the cell below.**


````python
def caesar(text, shift):
    result = []
    for ch in text:
        if "a" <= ch <= "z":
            result.append(chr((ord(ch) - ord("a") + shift) % 26 + ord("a")))
        elif "A" <= ch <= "Z":
            result.append(chr((ord(ch) - ord("A") + shift) % 26 + ord("A")))
        else:
            result.append(ch)
    return "".join(result)

print(caesar("Hello World", 3))
````

````python
class IDIterator:
    def __init__(self, n):
        self.current = 1
        self.n = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.n:
            raise StopIteration
        value = self.current
        self.current += 1
        return value

for value in IDIterator(5):
    print(value)
````

````python
from functools import wraps
from time import perf_counter

def log_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = perf_counter()
        result = func(*args, **kwargs)
        elapsed = perf_counter() - start
        print(f"{func.__name__}: {elapsed:.6f}s")
        return result
    return wrapper

@log_time
def add(a, b):
    return a + b

print(add(10, 20))
````

````python
from functools import wraps

def log(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Calling", func.__name__)
        return func(*args, **kwargs)
    return wrapper

def validate(func):
    @wraps(func)
    def wrapper(value):
        if value < 0:
            raise ValueError("Value must be non-negative")
        return func(value)
    return wrapper

@log
@validate
def square(value):
    return value * value

print(square(5))
````

### Question 5/5

**Problem:** Build a retry decorator that reruns a function a limited number of times when it raises an exception.

**Write the complete Python solution in the cell below.**


````python
from functools import wraps

def retry(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            last_error = None
            for _ in range(times):
                try:
                    return func(*args, **kwargs)
                except Exception as exc:
                    last_error = exc
            raise last_error
        return wrapper
    return decorator

attempts = 0

@retry(3)
def unstable():
    global attempts
    attempts += 1
    if attempts < 3:
        raise RuntimeError("temporary failure")
    return "success"

print(unstable())
````

### Question 1/5

**Problem:** Create a BankAccount class with deposit, withdrawal, and balance methods.

**Write the complete Python solution in the cell below.**


````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0 or amount > self.balance:
            raise ValueError("Invalid withdrawal")
        self.balance -= amount

    def get_balance(self):
        return self.balance

account = BankAccount("Yogesh", 10000)
account.deposit(2000)
account.withdraw(1500)
print(account.get_balance())
````

### Question 3/5

**Problem:** Build an inventory manager using OOP, dictionaries, file persistence, validation, and custom exceptions.

**Write the complete Python solution in the cell below.**


````python
class InvalidStockError(Exception):
    pass

class Inventory:
    def __init__(self):
        self.items = {}

    def add(self, sku, name, quantity):
        if quantity < 0:
            raise InvalidStockError("Quantity cannot be negative")
        self.items[sku] = {"name": name, "quantity": quantity}

    def sell(self, sku, quantity):
        if sku not in self.items:
            raise KeyError("Unknown SKU")
        if quantity > self.items[sku]["quantity"]:
            raise InvalidStockError("Insufficient stock")
        self.items[sku]["quantity"] -= quantity

inventory = Inventory()
inventory.add("P101", "Laptop", 10)
inventory.sell("P101", 2)

with open("inventory.txt", "w", encoding="utf-8") as f:
    for sku, item in inventory.items.items():
        f.write(f"{sku},{item['name']},{item['quantity']}\n")

print(inventory.items)
````

### Question 4/5

**Problem:** Build a student-result processor that reads records, calculates grades, ranks students, and exports a report.

**Write the complete Python solution in the cell below.**


````python
students = [
    {"name": "A", "marks": [80, 75, 90]},
    {"name": "B", "marks": [65, 70, 60]},
    {"name": "C", "marks": [92, 88, 95]}
]

def grade(avg):
    if avg >= 90: return "A"
    if avg >= 75: return "B"
    if avg >= 60: return "C"
    return "D"

for student in students:
    student["total"] = sum(student["marks"])
    student["average"] = student["total"] / len(student["marks"])
    student["grade"] = grade(student["average"])

students.sort(key=lambda s: s["average"], reverse=True)

with open("student_report.txt", "w", encoding="utf-8") as f:
    for rank, s in enumerate(students, 1):
        f.write(f"{rank}. {s['name']} {s['average']:.2f} {s['grade']}\n")

print(students)
````
